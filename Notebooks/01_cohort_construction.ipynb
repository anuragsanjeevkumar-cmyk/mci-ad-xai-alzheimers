```python
# Create diagnosis progression for each participant

def classify_progression(x):
    diagnoses = set(x.dropna())

    if 2.0 in diagnoses and 3.0 in diagnoses:
        return "MCI_to_dementia"
    elif 2.0 in diagnoses:
        return "MCI_only"
    else:
        return "Other"


progression = df.groupby("PTID")["DIAGNOSIS"].apply(classify_progression)

print("Participant progression:")
print(progression.value_counts())
```

```python
# Get the first MCI date for each participant
baseline_dates = first_mci[["PTID", "EXAMDATE"]].rename(
    columns={"EXAMDATE": "MCI_DATE"}
)

# Get all AD-positive observations
ad_events = df[df["DXAD"] == 1][
    ["PTID", "EXAMDATE", "VISCODE", "DIAGNOSIS", "DXAD"]
].copy()

# Add each participant's first MCI date
ad_events = ad_events.merge(
    baseline_dates,
    on="PTID",
    how="inner"
)

# Keep only AD events occurring AFTER first MCI
valid_ad_events = ad_events[
    ad_events["EXAMDATE"] > ad_events["MCI_DATE"]
].copy()

# Unique participants with a valid later AD event
valid_converter_ids = valid_ad_events["PTID"].unique()

print("Baseline MCI participants:",
      len(baseline_mci_ids))

print("Potential converters before time check:",
      len(baseline_mci_converters))

print("Converters with AD occurring AFTER first MCI:",
      len(valid_converter_ids))

print("Remaining potential non-converters:",
      len(baseline_mci_ids) - len(valid_converter_ids))
```

```python
# Get all diagnostic records from the original DXSUM data
# for the 174 participants identified as MCI -> AD converters.

converter_diagnostics = df[
    df["PTID"].isin(valid_converter_ids)
].copy()

# Diagnostic variables relevant to distinguishing AD from other dementias
diagnostic_cols = [
    "PTID",
    "EXAMDATE",
    "VISCODE",
    "DIAGNOSIS",
    "DXAD",
    "DXAPP",
    "DXAPROB",
    "DXAPOSS",
    "DXOTHDEM",
    "DXODES",
    "DXPARK",
    "DXPCOG",
    "DXDEP"
]

# Keep only columns that exist in the dataset
diagnostic_cols = [
    col for col in diagnostic_cols
    if col in converter_diagnostics.columns
]

print("Number of diagnostic records:", len(converter_diagnostics))

print("\nDiagnostic variables available:")
print(diagnostic_cols)

print("\nValues of diagnostic indicators among converter records:")
for col in diagnostic_cols[4:]:
    print(f"\n{col}:")
    print(converter_diagnostics[col].value_counts(dropna=False))
```

```python
# Get the first AD-related event after the participant's first MCI date

first_ad_event = (
    valid_ad_events
    .sort_values(["PTID", "EXAMDATE"])
    .drop_duplicates("PTID", keep="first")
    .copy()
)

print("Unique converter participants:", first_ad_event["PTID"].nunique())

print("\nFirst AD event records:")
display(
    first_ad_event[
        ["PTID", "MCI_DATE", "EXAMDATE", "VISCODE",
         "DIAGNOSIS", "DXAD"]
    ].head(15)
)

print("\nTime from first MCI to first AD event:")
time_to_ad = (
    first_ad_event["EXAMDATE"] - first_ad_event["MCI_DATE"]
).dt.days / 365.25

print(time_to_ad.describe())
```

```python
# Calculate the last available diagnostic follow-up date
# for each baseline MCI participant.

baseline_mci = first_mci[
    ["PTID", "EXAMDATE"]
].rename(columns={"EXAMDATE": "MCI_DATE"}).copy()

# Get all dated diagnostic visits for baseline MCI participants
followup = df[
    df["PTID"].isin(baseline_mci_ids) &
    df["EXAMDATE"].notna()
][["PTID", "EXAMDATE", "DIAGNOSIS"]].copy()

# Last observed diagnostic date for each participant
last_followup = (
    followup
    .groupby("PTID")["EXAMDATE"]
    .max()
    .reset_index()
    .rename(columns={"EXAMDATE": "LAST_DATE"})
)

# Add baseline MCI date
followup_summary = baseline_mci.merge(
    last_followup,
    on="PTID",
    how="left"
)

# Follow-up duration in years
followup_summary["FOLLOWUP_YEARS"] = (
    followup_summary["LAST_DATE"] -
    followup_summary["MCI_DATE"]
).dt.days / 365.25

# Identify converters
followup_summary["CONVERTER"] = (
    followup_summary["PTID"].isin(valid_converter_ids)
)

print("Overall follow-up duration:")
display(followup_summary["FOLLOWUP_YEARS"].describe())

print("\nFollow-up duration for converters:")
display(
    followup_summary.loc[
        followup_summary["CONVERTER"],
        "FOLLOWUP_YEARS"
    ].describe()
)

print("\nFollow-up duration for potential non-converters:")
display(
    followup_summary.loc[
        ~followup_summary["CONVERTER"],
        "FOLLOWUP_YEARS"
    ].describe()
)
```

```python
# Define a 2-year prediction horizon

HORIZON = 2.0

# Determine whether each participant converted to AD
# within 2 years of their first MCI visit.

first_ad_times = first_ad_event[
    ["PTID", "MCI_DATE", "EXAMDATE"]
].copy()

first_ad_times["TIME_TO_AD_YEARS"] = (
    first_ad_times["EXAMDATE"] -
    first_ad_times["MCI_DATE"]
).dt.days / 365.25

# Participants who converted within 2 years
converter_2yr_ids = set(
    first_ad_times.loc[
        first_ad_times["TIME_TO_AD_YEARS"] <= HORIZON,
        "PTID"
    ]
)

# Participants with no AD conversion within 2 years
# AND at least 2 years of follow-up
nonconverter_2yr_ids = set(
    followup_summary.loc[
        (~followup_summary["PTID"].isin(valid_converter_ids)) &
        (followup_summary["FOLLOWUP_YEARS"] >= HORIZON),
        "PTID"
    ]
)

# Participants excluded because they don't have enough follow-up
excluded_short_followup_ids = set(
    followup_summary.loc[
        (~followup_summary["PTID"].isin(valid_converter_ids)) &
        (followup_summary["FOLLOWUP_YEARS"] < HORIZON),
        "PTID"
    ]
)

print("2-year converters:", len(converter_2yr_ids))
print("2-year non-converters:", len(nonconverter_2yr_ids))
print("Excluded due to <2 years follow-up:", len(excluded_short_followup_ids))

print("\nTotal usable participants:",
      len(converter_2yr_ids) + len(nonconverter_2yr_ids))

print("\nClass distribution:")
print(
    "Converters:",
    len(converter_2yr_ids)
)
print(
    "Non-converters:",
    len(nonconverter_2yr_ids)
)
```

```python
# ---------------------------------------------------------
# Final 2-year MCI -> AD outcome definition
# ---------------------------------------------------------

HORIZON = 2.0

# First AD event for each participant
first_ad_times = first_ad_event[
    ["PTID", "MCI_DATE", "EXAMDATE"]
].copy()

first_ad_times["TIME_TO_AD_YEARS"] = (
    first_ad_times["EXAMDATE"] -
    first_ad_times["MCI_DATE"]
).dt.days / 365.25

# All baseline MCI participants with at least 2 years follow-up
eligible_2yr_ids = set(
    followup_summary.loc[
        followup_summary["FOLLOWUP_YEARS"] >= HORIZON,
        "PTID"
    ]
)

# Participants who converted within 2 years
converter_2yr_ids = set(
    first_ad_times.loc[
        first_ad_times["TIME_TO_AD_YEARS"] <= HORIZON,
        "PTID"
    ]
)

# Everyone with >=2 years follow-up who did NOT convert within 2 years
nonconverter_2yr_ids = (
    eligible_2yr_ids - converter_2yr_ids
)

# Participants with insufficient follow-up
excluded_short_followup_ids = (
    set(baseline_mci_ids) - eligible_2yr_ids
)

print("===== FINAL 2-YEAR COHORT CHECK =====")
print("Baseline MCI participants:", len(baseline_mci_ids))
print()
print("2-year converters:", len(converter_2yr_ids))
print("2-year non-converters:", len(nonconverter_2yr_ids))
print("Excluded (<2 years follow-up):", len(excluded_short_followup_ids))
print()
print("Total usable participants:",
      len(converter_2yr_ids) + len(nonconverter_2yr_ids))

print("\nExpected class ratio:")
print(
    "Converter percentage:",
    round(
        len(converter_2yr_ids) /
        (len(converter_2yr_ids) + len(nonconverter_2yr_ids)) * 100,
        2
    ),
    "%"
)
```

```python
# ---------------------------------------------------------
# Construct the final participant-level 2-year cohort
# ---------------------------------------------------------

# Select only the baseline MCI participants in our final cohort
final_cohort = first_mci[
    first_mci["PTID"].isin(
        converter_2yr_ids | nonconverter_2yr_ids
    )
].copy()

# Rename the baseline examination date
final_cohort = final_cohort.rename(
    columns={"EXAMDATE": "MCI_DATE"}
)

# Create the binary outcome
final_cohort["AD_CONVERSION_2YR"] = (
    final_cohort["PTID"].isin(converter_2yr_ids)
).astype(int)

# Keep only the variables we need at this stage
final_cohort = final_cohort[
    ["PTID", "MCI_DATE", "AD_CONVERSION_2YR"]
].copy()

# Sort and reset the index
final_cohort = (
    final_cohort
    .sort_values("PTID")
    .reset_index(drop=True)
)

print("===== FINAL COHORT =====")
print("Number of participants:", len(final_cohort))

print("\nOutcome distribution:")
print(final_cohort["AD_CONVERSION_2YR"].value_counts())

print("\nMissing values:")
print(final_cohort.isna().sum())

print("\nFirst 10 participants:")
display(final_cohort.head(10))
```

```python
# Extract baseline clinical measurements for the 997-person cohort

# Load the clinical datasets
adas = pd.read_csv("../Data/ADAS_from_R.csv")
cdr = pd.read_csv("../Data/CDR_from_R.csv")
mmse = pd.read_csv("../Data/MMSE_from_R.csv")
faq = pd.read_csv("../Data/FAQ_from_R.csv")

# Convert dates
for df in [adas, cdr, mmse, faq]:
    df["VISDATE"] = pd.to_datetime(df["VISDATE"])

# Make sure cohort date is datetime
final_cohort["MCI_DATE"] = pd.to_datetime(final_cohort["MCI_DATE"])


def get_baseline_measurements(df, value_columns):
    """
    Select the latest available measurement on or before
    each participant's first MCI date.
    """

    temp = df[
        ["PTID", "RID", "VISDATE"] + value_columns
    ].copy()

    # Attach MCI baseline date
    temp = temp.merge(
        final_cohort[["PTID", "MCI_DATE"]],
        on="PTID",
        how="inner"
    )

    # Only information available at or before prediction time
    temp = temp[
        temp["VISDATE"] <= temp["MCI_DATE"]
    ].copy()

    # Latest available assessment
    temp = (
        temp
        .sort_values(["PTID", "VISDATE"])
        .drop_duplicates("PTID", keep="last")
        .reset_index(drop=True)
    )

    return temp


# Extract baseline values
baseline_adas = get_baseline_measurements(
    adas,
    ["TOTSCORE", "TOTAL13"]
)

baseline_cdr = get_baseline_measurements(
    cdr,
    ["CDGLOBAL", "CDRSB"]
)

baseline_mmse = get_baseline_measurements(
    mmse,
    ["MMSCORE"]
)

baseline_faq = get_baseline_measurements(
    faq,
    ["FAQTOTAL"]
)


# Keep only the variables we need
baseline_adas = baseline_adas[
    ["PTID", "TOTSCORE", "TOTAL13"]
]

baseline_cdr = baseline_cdr[
    ["PTID", "CDGLOBAL", "CDRSB"]
]

baseline_mmse = baseline_mmse[
    ["PTID", "MMSCORE"]
]

baseline_faq = baseline_faq[
    ["PTID", "FAQTOTAL"]
]


# Merge everything into the final cohort
clinical_baseline = final_cohort.merge(
    baseline_adas, on="PTID", how="left"
)

clinical_baseline = clinical_baseline.merge(
    baseline_cdr, on="PTID", how="left"
)

clinical_baseline = clinical_baseline.merge(
    baseline_mmse, on="PTID", how="left"
)

clinical_baseline = clinical_baseline.merge(
    baseline_faq, on="PTID", how="left"
)


# Report availability
feature_columns = [
    "TOTSCORE",
    "TOTAL13",
    "CDGLOBAL",
    "CDRSB",
    "MMSCORE",
    "FAQTOTAL"
]

availability = pd.DataFrame({
    "Feature": feature_columns,
    "Available": [
        clinical_baseline[col].notna().sum()
        for col in feature_columns
    ],
    "Missing": [
        clinical_baseline[col].isna().sum()
        for col in feature_columns
    ]
})

availability["Availability_%"] = (
    availability["Available"] / len(clinical_baseline) * 100
).round(2)

print("Clinical baseline dataset shape:")
print(clinical_baseline.shape)

print("\nFeature availability:")
print(availability.to_string(index=False))
```

```python
import pandas as pd

# Load the final 997-participant cohort
cohort = pd.read_csv("../Data/final_cohort.csv")

# Load previously extracted baseline clinical tables
cdr = pd.read_csv("../Data/CDR_from_R.csv", low_memory=False)
mmse = pd.read_csv("../Data/MMSE_from_R.csv", low_memory=False)

# Convert dates
cohort["MCI_DATE"] = pd.to_datetime(cohort["MCI_DATE"])
cdr["VISDATE"] = pd.to_datetime(cdr["VISDATE"], errors="coerce")
mmse["VISDATE"] = pd.to_datetime(mmse["VISDATE"], errors="coerce")


# Function to select the latest measurement before/on first MCI
def get_baseline(df, value_column):

    merged = df.merge(
        cohort[["PTID", "MCI_DATE"]],
        on="PTID",
        how="inner"
    )

    before_mci = merged[
        merged["VISDATE"].notna() &
        (merged["VISDATE"] <= merged["MCI_DATE"])
    ].copy()

    baseline = (
        before_mci
        .sort_values(["PTID", "VISDATE"])
        .drop_duplicates("PTID", keep="last")
    )

    return baseline[["PTID", value_column]]


# Extract baseline values
cdr_base = get_baseline(cdr, "CDGLOBAL")
cdrsb_base = get_baseline(cdr, "CDRSB")
mmse_base = get_baseline(mmse, "MMSCORE")


# Create clinical feature table
clinical_base = cohort[["PTID"]].copy()

clinical_base = clinical_base.merge(
    cdr_base,
    on="PTID",
    how="left"
)

clinical_base = clinical_base.merge(
    cdrsb_base,
    on="PTID",
    how="left"
)

clinical_base = clinical_base.merge(
    mmse_base,
    on="PTID",
    how="left"
)


# Treat undocumented -1 as missing
clinical_base["CDGLOBAL"] = clinical_base["CDGLOBAL"].replace(
    -1, pd.NA
)


print("Clinical baseline table shape:", clinical_base.shape)

print("\nMissing values:")
print(clinical_base.isna().sum())

print("\nCDR Global distribution:")
print(
    clinical_base["CDGLOBAL"]
    .value_counts(dropna=False)
    .sort_index()
)

print("\nCDR-SB summary:")
print(clinical_base["CDRSB"].describe())

print("\nMMSE summary:")
print(clinical_base["MMSCORE"].describe())
```

```python
import pandas as pd

# -----------------------------
# 1. Load cohort
# -----------------------------
cohort = pd.read_csv("../Data/final_cohort.csv")

# -----------------------------
# 2. Load demographic data
# -----------------------------
demo = pd.read_csv(
    "../Data/PTDEMOG_demographics_from_R.csv"
)

cohort["MCI_DATE"] = pd.to_datetime(cohort["MCI_DATE"])
demo["VISDATE"] = pd.to_datetime(
    demo["VISDATE"],
    errors="coerce"
)

# Match demographic records before/on first MCI
demo_merge = demo.merge(
    cohort[["PTID", "MCI_DATE"]],
    on="PTID",
    how="inner"
)

demo_before = demo_merge[
    demo_merge["VISDATE"].notna() &
    (demo_merge["VISDATE"] <= demo_merge["MCI_DATE"])
].copy()

demo_base = (
    demo_before
    .sort_values(["PTID", "VISDATE"])
    .drop_duplicates("PTID", keep="last")
)

demo_base = demo_base[
    [
        "PTID",
        "RID",
        "PTGENDER",
        "PTEDUCAT",
        "PTETHCAT",
        "PTRACCAT"
    ]
]

# -----------------------------
# 3. Load ADSL
# -----------------------------
adsl = pd.read_csv(
    "../Data/ADSL_from_R.csv"
)

adsl["SUBJID"] = pd.to_numeric(
    adsl["SUBJID"],
    errors="coerce"
)

demo_base["RID"] = pd.to_numeric(
    demo_base["RID"],
    errors="coerce"
)

# Add Age + APOE
demo_base = demo_base.merge(
    adsl[["SUBJID", "AGE", "APOE"]],
    left_on="RID",
    right_on="SUBJID",
    how="left"
)

# -----------------------------
# 4. Load clinical features
# -----------------------------
cdr = pd.read_csv(
    "../Data/CDR_from_R.csv",
    low_memory=False
)

mmse = pd.read_csv(
    "../Data/MMSE_from_R.csv",
    low_memory=False
)

cdr["VISDATE"] = pd.to_datetime(
    cdr["VISDATE"],
    errors="coerce"
)

mmse["VISDATE"] = pd.to_datetime(
    mmse["VISDATE"],
    errors="coerce"
)


def get_baseline(df, value_column):

    merged = df.merge(
        cohort[["PTID", "MCI_DATE"]],
        on="PTID",
        how="inner"
    )

    before = merged[
        merged["VISDATE"].notna() &
        (merged["VISDATE"] <= merged["MCI_DATE"])
    ].copy()

    base = (
        before
        .sort_values(["PTID", "VISDATE"])
        .drop_duplicates("PTID", keep="last")
    )

    return base[["PTID", value_column]]


cdr_global = get_baseline(
    cdr,
    "CDGLOBAL"
)

cdr_sb = get_baseline(
    cdr,
    "CDRSB"
)

mmse_base = get_baseline(
    mmse,
    "MMSCORE"
)

# -----------------------------
# 5. Combine everything
# -----------------------------
primary = cohort[
    ["PTID", "MCI_DATE", "AD_CONVERSION_2YR"]
].copy()

primary = primary.merge(
    demo_base[
        [
            "PTID",
            "AGE",
            "PTGENDER",
            "PTEDUCAT",
            "PTETHCAT",
            "PTRACCAT",
            "APOE"
        ]
    ],
    on="PTID",
    how="left"
)

primary = primary.merge(
    cdr_global,
    on="PTID",
    how="left"
)

primary = primary.merge(
    cdr_sb,
    on="PTID",
    how="left"
)

primary = primary.merge(
    mmse_base,
    on="PTID",
    how="left"
)

# Treat undocumented -1 CDR Global as missing
primary["CDGLOBAL"] = primary["CDGLOBAL"].replace(
    -1,
    pd.NA
)

# Save
primary.to_csv(
    "../Data/primary_features.csv",
    index=False
)

print("Primary feature table created!")
print("Shape:", primary.shape)

print("\nColumns:")
print(primary.columns.tolist())

print("\nMissing values:")
print(primary.isna().sum())
```

```python
import pandas as pd

from sklearn.model_selection import train_test_split

# Load final dataset
df = pd.read_csv(
    "../Data/final_modeling_dataset.csv"
)

# -----------------------------
# Primary feature set
# -----------------------------

primary_features = [
    "AGE",
    "PTGENDER",
    "PTEDUCAT",
    "PTETHCAT",
    "PTRACCAT",
    "APOE",
    "CDGLOBAL",
    "CDRSB",
    "MMSCORE"
]

X = df[primary_features].copy()
y = df["AD_CONVERSION_2YR"].copy()

# -----------------------------
# Stratified train/test split
# -----------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

print("Dataset split")
print("=" * 50)

print("Total participants:", len(df))
print("Training participants:", len(X_train))
print("Testing participants:", len(X_test))

print("\nTraining outcome:")
print(y_train.value_counts())

print("\nTraining outcome percentage:")
print(
    (y_train.value_counts(normalize=True) * 100)
    .round(2)
)

print("\nTesting outcome:")
print(y_test.value_counts())

print("\nTesting outcome percentage:")
print(
    (y_test.value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Numerical features
numeric_features = [
    "AGE",
    "PTEDUCAT",
    "CDGLOBAL",
    "CDRSB",
    "MMSCORE"
]

# Categorical features
categorical_features = [
    "PTGENDER",
    "PTETHCAT",
    "PTRACCAT",
    "APOE"
]

# Numerical preprocessing
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

# Categorical preprocessing
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        ))
    ]
)

# Combine preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

# Fit ONLY on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform test data using training-fitted preprocessing
X_test_processed = preprocessor.transform(X_test)

print("Preprocessing completed")
print("=" * 50)

print("Original training shape:", X_train.shape)
print("Processed training shape:", X_train_processed.shape)

print("Original testing shape:", X_test.shape)
print("Processed testing shape:", X_test_processed.shape)

print("\nProcessed data type:")
print(X_train_processed.dtype)
```
